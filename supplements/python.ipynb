# 可变与不可变

```python
d = {}
print(f"字典初始地址: {id(d)}")

# 修改字典：添加一个键值对
d[0] = [0]
print(f"修改后字典值: {d}")
print(f"修改后字典地址: {id(d)}")

# 结论：地址完全一样，说明是在原内存地址上进行的修改
print(f"地址是否相同: {id(d) == id(d)}") 
```

```python
l = [1, 2, 3]
print(f"列表初始地址: {id(l)}")

# 修改列表：追加元素
l.append(4)
# 修改列表：修改索引处的值
l[0] = 99

print(f"修改后列表值: {l}")
print(f"修改后列表地址: {id(l)}")

# 验证：地址不变
```

```python
s = {1, 2, 3}
print(f"集合初始地址: {id(s)}")

# 修改集合：添加元素
s.add(4)
# 修改集合：移除元素
s.remove(1)

print(f"修改后集合值: {s}")
print(f"修改后集合地址: {id(s)}")

# 验证：地址不变
```

```python
s = "Hello"
old_id = id(s)
s = s + " World" # 这实际上是创建了一个全新的字符串对象，并让变量 s 指向它
new_id = id(s)

print(f"字符串修改前地址: {old_id}")
print(f"字符串修改后地址: {new_id}")
print(f"地址是否变化: {old_id != new_id}") # True，地址变了！
```

```python
print("--- 不可变类型演示 ---")

# 数字 (int)
a = 100
print(f"数字初始值: {a}, 地址: {id(a)}")
a = a + 1
print(f"数字修改后: {a}, 地址: {id(a)}") 
# 结论：地址变了，说明旧的 100 没变，只是 a 指向了新的 101

# 字符串 (str)
s = "Hello"
print(f"字符串初始: {s}, 地址: {id(s)}")
s += " World"
print(f"字符串修改: {s}, 地址: {id(s)}")
# 结论：地址变了，Python 销毁了旧字符串（或等待回收），创建了新字符串
```

```python
t = (1, 2, [3, 4])
print(f"元组初始内容: {t}, 地址: {id(t)}")

try:
    t[0] = 99  # 尝试修改元素
except TypeError as e:
    print(f"修改尝试失败: {e}")

# 特殊情况：元组内的“可变对象”
t[2].append(5) 
print(f"修改元组内的列表后: {t}, 地址: {id(t)}")
# 结论：元组本身的地址没变，它对列表的“引用”没变，变的是列表内部。
# 但元组依然被视为不可变容器。
```

# 深拷贝和浅拷贝

浅拷贝：重新分配一块内存，创建一个新的对象，但里面的元素是原对象中各个子对象的引用。

深拷贝：重新分配一块内存，创建一个新的对象，并且递归地复制原对象中的所有子对象。

```python
import copy

# 1. 构造一个嵌套列表（原对象）
# 内部包含一个不可变对象 (str) 和一个可变对象 (list)
original = ["Python", [1, 2, 3]]

# 2. 执行浅拷贝 (Shallow Copy)
shallow_copy = copy.copy(original)

# 3. 执行深拷贝 (Deep Copy)
deep_copy = copy.deepcopy(original)

print(f"原始对象: \t{original}，地址: {id(original)}")
print(f"浅拷贝: \t\t{shallow_copy}，地址: {id(shallow_copy)}（改变）")
print(f"深拷贝: \t\t{deep_copy}，地址: {id(deep_copy)}（改变）")
print("-" * 50)

# --- 核心实验：修改嵌套的可变对象 ---
print("修改原始对象中的子列表: original[1].append(4)")
original[1].append(4)

print(f"原始对象: \t{original}")
# 浅拷贝被影响了，因为它的子对象引用没变
print(f"浅拷贝: \t\t{shallow_copy}（改变）")
# 深拷贝没受影响，因为它是完全独立的副本
print(f"深拷贝: \t\t{deep_copy}（相同）")

print("-" * 50)
# --- 验证子对象的内存地址 ---
print(f"原始子列表地址: \t\t{id(original[1])}")
print(f"浅拷贝子列表地址: \t{id(shallow_copy[1])} (相同)")
print(f"深拷贝子列表地址: \t{id(deep_copy[1])} (改变)")
```

```python
class MyResourceManager:
    def __init__(self, name):
        self.name = name
        print(f"[1. __init__]: 实例已创建，名字是 {self.name}")

    def __enter__(self):
        print(f"[2. __enter__]: 进入 with 代码块之前执行。")
        # 重点：这里的返回值会赋给 as 后面的变量
        return f"【我是 __enter__ 的返回值，我被赋给了 as 后的变量】"

    def __exit__(self, exc_type, exc_val, exc_tb):
        print(f"[4. __exit__]: with 代码块执行完毕，现在清理资源。")
        # exc_type, exc_val, exc_tb 用于处理异常，这里暂不深入
        return False

# --- 测试执行 ---
print("--- 开始运行 with 语句 ---")

with MyResourceManager("测试资源") as resource:
    print(f"[3. 内部执行]: 拿到资源了 -> {resource}")
    print("   正在处理业务逻辑...")

print("--- with 语句结束 ---")
```

```python
class Robot:
    def __new__(cls, *args, **kwargs):
        """
        1. __new__ 是真正的构造方法，负责创建实例。
        它是静态方法（虽然不需要显式写 @staticmethod，但行为上是）。
        这个时候还没有对象，所以没有self
        __new__接收类对象cls，根据类对象创建实例
        """
        print(f"\n[1. __new__ 阶段]: 正在为 {cls.__name__} 分配内存空间...")
        print(f"   __new__ 接收到的参数: args={args}, kwargs={kwargs}")
        
        # 调用父类 object 的 __new__ 来实际创建对象
        instance = super().__new__(cls)
        
        # 验证：__new__ 确实拿到了传入的参数，以便决定如何创建对象
        print(f"   创建出的实例内存地址: {id(instance)}")
        
        # 关键：必须返回这个实例，否则 __init__ 不会被调用
        return instance

    def __init__(self, name, version):
        """
        2. __init__ 是初始化方法，负责给已经创建好的实例“穿衣服”（设置属性）。
        这里的 self 就是 __new__ 返回的那个 instance。
        """
        print(f"[2. __init__ 阶段]: 正在初始化实例属性...")
        print(f"   __init__ 接收到的 self 地址: {id(self)}")
        
        self.name = name
        self.version = version
        print(f"   属性设置完成: name={self.name}, version={self.version}")

# --- 执行实例化 ---
# 当我们调用 Robot(...) 时，*args 和 **kwargs 会同时传给 __new__ 和 __init__
my_robot = Robot("机器人", version="v2.0")
```

# Wrapper

```python
import time
import functools

# 1. 定义装饰器：它是一个高阶函数
def timer_decorator(func): # 把func的函数名指向wrapper
    """
    这是一个统计运行时间的装饰器
    func: 被装饰的函数对象
    """
    cross_invoke_param = []
    # 2. 定义嵌套函数（包装器）：它负责添加新功能
    @functools.wraps(func) # 把 func 的属性拷贝到了 wrapper 上。
    def wrapper(*args, **kwargs):
        # --- 新功能：开始前记录时间 ---
        start_time = time.time()
        print(f"[日志] 准备执行函数: {func.__name__}")
        
        # 3. 执行原函数：使用解包语法传递参数
        result = func(*args, **kwargs)
        
        # --- 新功能：结束后计算耗时 ---
        end_time = time.time()
        cross_invoke_param.append(end_time - start_time)
        print(f"[日志] 执行完毕，耗时: {end_time - start_time:.4f}s")
        
        # 4. 返回原函数的执行结果
        return result
    
    # 5. 返回包装后的函数对象
    return wrapper

# --- 使用方式 ---

@timer_decorator
def complex_calculation(n):
    """模拟一个耗时的计算任务"""
    total = 0
    for i in range(n):
        total += i
    time.sleep(0.5) # 模拟延迟
    return total

# 调用方式完全不变
res = complex_calculation(1)
print(f"计算结果: {res}")
```

# 高阶函数

```python
import inspect

class Klass:
    def class_method(self):
        self.class_param = [1, 2]  # 实例属性存储在self的属性字典中
        local_param = {"key": "original"} # 局部变量储存在class_method的闭包中
        
        print(f"--- 初始状态 ---")
        print(f"class_param id: {id(self.class_param)}")
        print(f"local_param id: {id(local_param)}")

        def inner_method():
            print(f"\n--- Inner Method ---")
            self.class_param.append(3) 
            print(f"class_param 修改: {id(self.class_param)}")
            self.class_param = ["a", "new", "list"]
            print(f"class_param 重新赋值: {id(self.class_param)}")

            # 注意：如果不写 nonlocal，下一行会创建一个同名的局部变量，而不是修改外部的 local_param
            nonlocal local_param
            local_param["newkey"] = "new key" 
            print(f"local_param 修改: {id(local_param)}")
            local_param = {"new_key": "new_value"} 
            print(f"local_param 重新赋值: {id(local_param)}")


        # 查看闭包信息
        inner_method()
        print(f"\n--- 执行 Inner 后 ---")
        print(f"self.class_param: {self.class_param}")
        print(f"local_param: {local_param}")
        
        # 验证闭包存储
        if inner_method.__closure__:
            cells = [cell.cell_contents for cell in inner_method.__closure__]
            print(f"\n闭包捕获的内容: {cells}")

obj = Klass()
obj.class_method()
```

```python
from functools import reduce

# 准备数据：一组员工信息（姓名, 年龄, 薪资）
employees = [
    {"name": "Alice", "age": 28, "salary": 5000},
    {"name": "Bob", "age": 35, "salary": 8000},
    {"name": "Charlie", "age": 22, "salary": 4000},
    {"name": "David", "age": 40, "salary": 12000},
    {"name": "Eve", "age": 30, "salary": 7000}
]

# --- 1. filter (过滤): 筛选出年龄大于 30 岁的员工 ---
# 逻辑：保留返回 True 的元素
# filter(function, iterable)
seniors = filter(lambda e: e["age"] > 30, employees) # iterator
print(f"Seniors: {list(seniors)}") 
# 输出: [{'name': 'Bob', ...}, {'name': 'David', ...}]


# --- 2. map (映射): 提取所有员工的薪资，并给每人涨 10% ---
# 逻辑：对每个元素执行函数并返回新值
# map(function, iterable)
raised_salaries = map(lambda e: e["salary"] * 1.1, employees)
# 注意：map 返回的是迭代器，我们转成列表查看
salaries_list = list(raised_salaries)
print(f"Raised Salaries: {salaries_list}")
# 输出: [5500.0, 8800.0, 4400.0, 13200.0, 7700.0]


# --- 3. reduce (归约): 计算涨薪后的总成本 ---
# 逻辑：f(f(f(a, b), c), d) 递归累加
# reduce(func, iterable, initializer)
total_cost = reduce(lambda x, y: x + y, salaries_list)
print(f"Total Cost: {total_cost}")
# 输出: 39600.0
```

# 匿名函数

lambda 函数是一种小型、匿名的、内联函数，它可以具有任意数量的参数，但只能有一个表达式。

lambda 函数通常用于编写简单的、单行的函数，通常在需要函数作为参数传递的情况下使用

```python
# 普通函数版本
def add(x, y):
    return x + y

# Lambda 版本
add_lambda = lambda x, y: x + y

print("add(5, 3) == add_lambda(5, 3)",add(5, 3) == add_lambda(5, 3))        # 输出 8


students = [("Alice", 88), ("Bob", 95), ("Charlie", 70)]

# 使用 lambda 指定排序的“键”为元组的第二个元素（分数）
students.sort(key=lambda s: s[1])

print(students) 
# 输出: [('Charlie', 70), ('Alice', 88), ('Bob', 95)]

def make_multiplier(n):
    # 返回一个 lambda 函数，它记住了 n 的值
    return lambda x: x * n

double = make_multiplier(2) # n 是 2
triple = make_multiplier(3) # n 是 3

print(double(10)) # 输出 20
print(triple(10)) # 输出 30

result = (lambda x, y: x ** y)(2, 3)
print(result) # 输出 8 (2的3次方)
```

# 迭代器和生成器

```python
# 1. 准备不同类型的可迭代对象
# 序列：列表、元组、字符串、range、byte和bytearray
# 集合：dict，set，frozenset
# 文件、生成器和迭代器
my_list = [10, 20]
my_str = "PY"
my_tuple = ("A", "B")

def demonstrate_iterator(collection):
    print(f"\n--- 正在演示集合类型: {type(collection).__name__} ---")
    
    # 2. 创建迭代器对象 (使用 iter() 方法)
    # 此时迭代器内部会初始化一个指针，指向集合的第一个元素之前
    it = iter(collection)
    print(f"迭代器对象已创建: {it}")

    # 3. 使用 next() 访问元素
    # 迭代器会记住当前位置，每次调用 next()，指针向后移动一位
    try:
        print(f"第一次调用 next(): {next(it)}")
        print(f"第二次调用 next(): {next(it)}")
        
        # 4. 证明“只能往前，不能后退”
        # 此时已经访问完所有元素，如果我们想再次访问第一个元素，
        # 除非重新创建迭代器，否则原迭代器无法“倒车”。
        
        print("尝试第三次调用 next()...")
        print(next(it)) 
    except StopIteration:
        # 5. 迭代结束标志
        # 当所有的元素被访问完，再次调用 next() 会抛出 StopIteration 异常
        print("捕获到 StopIteration: 所有元素已访问完毕，迭代器无法后退或重置。")

# 演示列表
demonstrate_iterator(my_list)

# 演示字符串
# demonstrate_iterator(my_str)

# 演示如何手动遍历（模拟 for 循环的本质）
print("\n--- 模拟 for 循环的底层实现 ---")
data = [1, 2, 3]
it = iter(data)
while True:
    try:
        item = next(it)
        print(f"处理元素: {item}")
    except StopIteration:
        print("循环结束")
        break
```

```python
def func2():
    print(">>> [生成器内部] 准备开始执行第一段逻辑")
    print('######')
    yield 100
    
    print(">>> [生成器内部] 准备开始执行第二段逻辑")
    print('------')
    yield 200
    
    print(">>> [生成器内部] 准备开始执行最后一段逻辑")
    print('======')
    return 300

# 1. 演示：调用函数不执行函数体
print("--- 步骤 1: 调用函数 ---")
gen = func2() # 返回生成器
print(f"函数已调用，返回的对象类型是: {type(gen)}")
# 此时你会发现，'######' 并没有被打印，说明函数体确实没运行

# 2. 演示：第一次 next()
print("\n--- 步骤 2: 第一次执行 next() ---")
val1 = next(gen)
print(f"接收到的 yield 值: {val1}")

# 3. 演示：第二次 next()，展示从断点继续
print("\n--- 步骤 3: 第二次执行 next() ---")
val2 = next(gen)
print(f"接收到的 yield 值: {val2}")

# 4. 演示：最后一次执行，处理 StopIteration 和 return 值
print("\n--- 步骤 4: 第三次执行 next() ---")
try:
    val3 = next(gen)
except StopIteration as e:
    # return 的值 300 会包含在 StopIteration 异常的 value 属性中
    print(f"生成器已结束，捕获到 StopIteration。Return 的值是: {e.value}")

```

```python
import numpy as np
import pandas as pd
import torch

def check_truth(label, value):
    """辅助函数：打印值的真假性"""
    print(f"{label:<15} | 值: {repr(value):<10} | 结果: {bool(value)}")

print(f"{'类型':<15} | {'原始值':<10} | 布尔值")
print("-" * 45)

# 1. 数值型 (Integer & Float)
check_truth("Int Zero", 0)
check_truth("Int Positive", 1)
check_truth("Int Negative", -1)
check_truth("Float Zero", 0.0)
check_truth("Float Positive", 1.0)
check_truth("Float Negative", -1.0)

# 2. 容器型 (String, List, Tuple, Dict, Set)
check_truth("Str Empty", "")
check_truth("Str Full", "abc")
check_truth("List Empty", [])
check_truth("List Full", [1])
check_truth("Tuple Empty", ())
check_truth("Tuple Full", (1, 2))
check_truth("Dict Empty", {})
check_truth("Dict Full", {"a": 1})
check_truth("Set Empty", set())
check_truth("Set Full", {1})

# 4. 自定义对象 (以链表节点为例)
class Node:
    def __init__(self, val=0):
        self.val = val
        self.next = None

empty_node = Node() 
check_truth("Custom Object", empty_node.__class__) # 默认情况下，对象只要存在就是 True

# 3. None 类型
check_truth("NoneType", None)

```

# 闭包

```python
# 定义一个全局变量
counter = 10

def modify_global():
    # 如果不加下面这一行，Python 会报错或创建一个局部变量 counter
    # global counter 
    
    print(f"函数内部访问全局变量: {counter}")
    counter += 1  # 修改全局变量
    print(f"函数内部修改后的全局变量: {counter}")

print(f"执行前全局变量的值: {counter}")
modify_global()
print(f"执行后全局变量的值: {counter}")
```

```python
def make_counter(start_value):
    # 这是一个外函数中的局部变量
    count = start_value

    def increment():
        # 闭包的核心：引用了外层函数的变量 count
        # 使用 nonlocal 声明：我们要修改的是外层嵌套作用域中的变量，而不是全局变量
        nonlocal count
        count += 1
        return count

    # 返回内函数的引用（注意：这里返回的是函数对象本身，没带括号）
    return increment

# 创建一个闭包实例
my_counter = make_counter(100)

print(f"第一次调用闭包: {my_counter()}") # 输出 101
print(f"第二次调用闭包: {my_counter()}") # 输出 102

# 证明闭包的独立性
another_counter = make_counter(0)
print(f"另一个独立闭包: {another_counter()}") # 输出 1
```

```python
var = "global"  # Global (全局变量)

def outer_func():
    var = "local"  # Enclosing (外层函数的局部变量)
    
    def inner_func():
        nonlocal var  # 绑定到 outer_func 里的 var
        print("nonlocal var =", var)

    def inner_func2():
        global var
        print(f"global var =", var)
    
    inner_func()
    inner_func2()

outer_func()
```

```python
# 好无聊的问题
# 就近原则
def first_func():
    var = "in first func"

    def second_func():
        var = "in second func"

        def third_func():
            nonlocal var
            print(var)
        
        third_func()
    
    second_func()

first_func()
```

```python
# 真的好无聊
def first_func():
    var = "in first func"

    def second_func():
        nonlocal var
        var = var + ", a tail"

        def third_func():
            nonlocal var
            print(var)
        
        third_func()
    
    second_func()

    print(var)

first_func()
```

```python
# 闭包延迟陷阱
lst = [lambda x: x*i for i in range(4)]
res = [m(2) for m in lst]
print(res) # [0,2,4,6]?
# 这里 i 存在 __closure__ 里，是一个引用，指向同一个内存地址。
print(lst[0].__closure__[0].cell_contents) # 输出 3

# 等价于
def func():
    fun_list = []
    for i in range(4):
        def foo(x):
            return x*i
        fun_list.append(foo)
    return fun_list

# 解决
lst = [lambda x, i=i: x*i for i in range(4)]
res = [m(2) for m in lst]
print(res) # [0,2,4,6]

# partial
# 偏函数是将所要承载的函数作为partial()函数的第一个参数，原函数的各个参数依次作为partial()函数后续的参数，除非使用关键字参数。
from functools import partial
lst = [partial(lambda x, i: x * i, i=i) for i in range(4)]
res = [m(2) for m in lst]
print(res) # [0,2,4,6]
```

# 垃圾回收

```python
import sys
import gc

# 辅助函数：打印分割线
def print_title(title):
    print(f"\n{'='*20} {title} {'='*20}")

# 1. 引用计数演示
print_title("1. 引用计数 (Reference Counting)")

class MyObject:
    pass

obj = MyObject()
print(f"创建对象后，引用计数: {sys.getrefcount(obj) - 1}") # -1 是因为 getrefcount 也会产生临时引用

a = obj
b = obj
print(f"增加两个引用后，引用计数: {sys.getrefcount(obj) - 1}")

del a
del b
print(f"删除两个引用后，引用计数: {sys.getrefcount(obj) - 1}")


# 2. 循环引用与标记清除演示
print_title("2. 标记清除 (Mark and Sweep)")

# 暂时关闭自动回收，手动演示
gc.disable()
# 如果不关闭gc，每当新创建的对象数减去被销毁的对象数的差值超过阈值（默认 700）时，就会自动触发一次 0 代 GC。
# 或者程序退出时触发

class Node:
    def __init__(self, name):
        self.name = name
        self.next = None
    def __del__(self):
        print(f"  [GC] {self.name} 被销毁了")

# 创建循环引用
node1 = Node("Node1")
node2 = Node("Node2")
node1.next = node2
node2.next = node1

print(f"删除变量前 Node1 引用数: {sys.getrefcount(node1) - 1}")

# 删除局部变量引用
del node1
del node2
# 此时引用计数不为0（因为它们互相指向），对象依然存在于内存中
print("变量已 del，但循环引用导致对象未被销毁，析构方法未调用。")

# 手动触发垃圾回收（执行标记清除）
print("手动执行 gc.collect()...")
unreachable_count = gc.collect() 
print(f"回收了 {unreachable_count} 个不可达对象。")


# 3. 分代回收演示
print_title("3. 分代回收 (Generational Collection)")

# 查看当前分代回收的阈值
# 默认通常是 (700, 10, 10) 
# 意为：当第0代对象超过700个，触发0代回收；
# 当0代回收触发10次，触发1代回收；1代触发10次，触发2代回收。
print(f"当前分代回收阈值: {gc.get_threshold()}")

# 查看各代当前对象数量
print(f"各代当前对象数: {gc.get_count()}")

# 模拟对象从 0 代晋升到 1 代
l = []
print("创建大量对象以触发第0代回收...")
for i in range(800):
    l.append([i]) # 创建一些列表对象

print(f"代当前对象数: {gc.get_count()}")
print("解释：存活下来的对象会从第0代移动到第1代，减少回收频率，从而提高效率。")

# 重新开启自动回收
gc.enable()

print(f"触发部分回收后的对象数: {gc.get_count()}")
```
